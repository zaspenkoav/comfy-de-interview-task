# Retail Analytics — Data Engineering Task

**Дані**: Таблиця `retail_transactions` (~10K рядків, омніканальний ритейлер електроніки)

**Канали**: online (e-commerce), offline (фізичні магазини), marketplace

**Час**: ~10 хвилин

```python
\!pip install -q duckdb

import duckdb

conn = duckdb.connect()  # in-memory
conn.execute("CREATE SCHEMA interview_tasks")
conn.execute("""
    CREATE TABLE interview_tasks.retail_transactions AS
    SELECT
        transaction_id,
        transaction_date::DATE AS transaction_date,
        transaction_timestamp::TIMESTAMP AS transaction_timestamp,
        channel,
        CASE WHEN store_id = '' THEN NULL ELSE CAST(store_id AS INT) END AS store_id,
        NULLIF(customer_id, '') AS customer_id,
        NULLIF(session_id, '') AS session_id,
        NULLIF(device_type, '') AS device_type,
        NULLIF(utm_source, '') AS utm_source,
        product_id,
        category,
        product_name,
        CAST(quantity AS INT) AS quantity,
        CAST(unit_price AS DECIMAL(12,2)) AS unit_price,
        CAST(total_amount AS DECIMAL(12,2)) AS total_amount,
        payment_method,
        CASE WHEN is_returned = 'True' THEN TRUE ELSE FALSE END AS is_returned
    FROM read_csv('data/retail_transactions.csv', header=true, all_varchar=true)
""")

conn.sql('SELECT COUNT(*) as rows FROM interview_tasks.retail_transactions').show()
print('Data ready\!')
```

```python
# Explore schema
conn.sql('DESCRIBE interview_tasks.retail_transactions').show()
```
